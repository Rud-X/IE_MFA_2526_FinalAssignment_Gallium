```python

# Load the csv "data/EUROSTAT_20260604.csv" into a DataFrame
eurostat_df = pd.read_csv("data/EUROSTAT_20260604.csv")

# Transform "OBS_VALUE" to numeric, coercing errors to NaN
eurostat_df["OBS_VALUE"] = pd.to_numeric(eurostat_df["OBS_VALUE"], errors="coerce")
eurostat_df["product"] = eurostat_df["product"].astype(str)
print("########## eurostat_df: ##########")
# display(eurostat_df)

# Load the excel file "data/ANNEXII_UpdateOfCodes_PRC2024_PRC2025.xlsx" into a DataFrame
code_update_df = pd.read_excel("data/ANNEXII_UpdateOfCodes_PRC2024_PRC2025.xlsx")
# Remove "." from the "PRODCOM2024" and "PRODCOM2025" columns and convert to string
code_update_df["PRODCOM2024"] = code_update_df["PRODCOM2024"].str.replace(".", "", regex=False).astype(str)
code_update_df["PRODCOM2025"] = code_update_df["PRODCOM2025"].str.replace(".", "", regex=False).astype(str)
print("########## code_update_df: ##########")
# display(code_update_df)

# Use code_update_df to update the "product" codes in eurostat_df, replacing old PRODCOM2024 codes with new PRODCOM2025 codes
code_update_dict = dict(zip(code_update_df["PRODCOM2024"].astype(str), code_update_df["PRODCOM2025"].astype(str)))
eurostat_df["product"] = eurostat_df["product"].replace(code_update_dict)
print("########## eurostat_df after code update: ##########")
# display(eurostat_df)

# Merge eurostat_df with prodcom_desc_df on "product" and "PRODCOM_CODE", keeping all rows from eurostat_df
eurostat_df = eurostat_df.merge(
    prodcom_desc_df[["PRCCODE", "DESCRIPTION"]],
    left_on="product",
    right_on="PRCCODE",
    how="left",
)

# Display the merged DataFrame
print("########## Merged eurostat_df with prodcom_desc_df: ##########")
# display(eurostat_df)

# Extract all "product" values where "DESCRIPTION" is missing value
missing_desc_products = eurostat_df[eurostat_df["DESCRIPTION"].isna()]["product"].unique()
if len(missing_desc_products) > 0:
    print("########## Products with missing DESCRIPTION: ##########")
    print(missing_desc_products)

# Create a dictionary that "product" to "DESCRIPTION" for all rows where "DESCRIPTION" is not missing value
product_desc_dict = dict(
    zip(eurostat_df["product"][eurostat_df["DESCRIPTION"].notna()], eurostat_df["DESCRIPTION"][eurostat_df["DESCRIPTION"].notna()])
)
# display(product_desc_dict)

# Product PRODCOM codes to include
filter_on_include = True
if filter_on_include:
    PRODCOM_codes_to_include = [
        "24453073", # Unwrought gallium; gallium powders
        # "24453074", # Gallium and articles thereof
        # "20595320", # Wafers excl. silicon
        # "26112220", # LEDs
        # "27902050", # Indicator panels incorporating LEDs
    ]
    product_desc_dict = {k: v for k, v in product_desc_dict.items() if k in PRODCOM_codes_to_include}
    display(product_desc_dict)

    # Filter eurostat_df to only include rows where "product" is in product_desc_dict keys
    eurostat_df = eurostat_df[eurostat_df["product"].isin(product_desc_dict.keys())]

display(eurostat_df)

# Display number of values in "OBS_VALUE" that are not NaN for each "product" and for each "indicators"
print("########## Count of non-NaN OBS_VALUE by product and indicators: ##########")
print(eurostat_df.groupby(["product", "indicators"])["OBS_VALUE"].count())

# Show as a bar plot the count of non-NaN OBS_VALUE by "product" and "indicators"
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
eurostat_df.groupby(["product", "indicators"])["OBS_VALUE"].count().unstack().plot(kind="bar", stacked=True)
plt.title("Count of non-NaN OBS_VALUE by Product and Indicators")
plt.xlabel("Product")
plt.ylabel("Count of non-NaN OBS_VALUE")
plt.xticks(rotation=45)
plt.legend(title="Indicators")
plt.tight_layout()
plt.show()
```
